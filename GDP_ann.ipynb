```python
import polars as pl
import pandas as pd
import numpy as np
import plotly.express as px

pl.Config.set_tbl_hide_column_data_types(True)

GDP_Ann_df = pl.read_csv("API_Download_DS2_EN_csv_v2_179699.csv")
GDP_Ann_df
```

```python
GDP_Ann_df = GDP_Ann_df.drop(["Country Code", "Indicator Name", "Indicator Code", ""])

year_columns = [str(year) for year in range(2000, 2023)]
GDP_Ann_df = GDP_Ann_df.with_columns([pl.col(col) / 1e9 for col in year_columns])
GDP_Ann_df = GDP_Ann_df.with_columns([pl.col(col).round() for col in year_columns])

GDP_Ann_df
```

```python
GDP_Ann_df.null_count()
```

```python
year_columns = [str(year) for year in range(2000, 2023)]



GDP_Ann_df_long = GDP_Ann_df.unpivot( index=["Country Name"],
    on=year_columns,
    variable_name="Year",
    value_name="GDP")

GDP_Ann_df_long = GDP_Ann_df_long.with_columns(pl.col("Year").cast(pl.Int32))
selected_countries = ["United States", "China", "Japan", "Germany", "United Kingdom"]
GDP_Ann_df_filtered = GDP_Ann_df_long.filter(pl.col("Country Name").is_in(selected_countries))

GDP_Ann_df_filtered
```

```python
fig = px.line(GDP_Ann_df_filtered, x="Year", y="GDP", color="Country Name", markers=True, 
title="GDP over time (2000 - 2022)across Wealthy Countries", 
labels={"GDP": "GDP (Billions USD)", "Year": "Year", "Country Name": "Country"})

fig.show()
```
